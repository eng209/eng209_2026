## Exercice 2: MyLittleFacebook

Considérez le code suivant:

````python
from collections import defaultdict

# MyLittleFacebook
friendships: defaultdict[str, set[str]] = defaultdict(set)

def add_friends(name1: str, name2: str) -> None:
    friendships[name1].add(name2)
    friendships[name2].add(name1)
        
add_friends("Alex", "Victor")
add_friends("Alex", "Emelyne")
add_friends("Alex", "Emelyne")
add_friends("Emelyne", "Rose")
````

Ce code initialise un dictionnaire qui relie un nom (sous forme simple de `str`) à un ensemble de noms représentant ses amis. La relation ainsi modélisée est symétrique: si $A$ est un ami de $B$, alors $B$ est un ami de $A$.

Remarquez que les règles d'accès (_scoping_) en Python permettent de ne pas passer `friendships` comme paramètre supplémentaire à la fonction `add_friends`, qui a automatiquement accès à ce qui a été défini dans les _scopes_ lexicales parentes.
