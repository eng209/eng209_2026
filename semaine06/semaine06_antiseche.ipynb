⚠️ Attention : reshape ne retourne pas une copie du tableau, mais une vue (ou perspective) sur les mêmes données.

➡️ Cela signifie que toute modification du tableau retourné peut également modifier le tableau d'origine (_flatten_ retourne une copie)

💡 Astuce: pour éviter cela, utilisez _.reshape(...).copy()_ afin de créer une copie indépendante des données.

```python
import numpy as np
from numpy.typing import NDArray

arr_1d: NDArray[np.int64] = np.arange(30)
arr_3d: NDArray[np.int64] = arr_1d.reshape(2, 3, 5)  # reshape crée une vue si possible

# On va comparer arr_1d[7] et arr_3d[0, 1, 2]
print(f"Array 1D de forme {arr_1d.shape}, valeur à l’indice [7] : {arr_1d[7]}")
print(f"Array 3D de forme {arr_3d.shape}, valeur à l’indice [0, 1, 2] : {arr_3d[0, 1, 2]}")

# Modification via l’array 3D
arr_3d[0, 1, 2] = 99

print(f"\nAprès modification de arr_s3[0, 1, 2]:")
print(f"arr_3d[0, 1, 2] = {arr_3d[0, 1, 2]}")
print(f"arr_1d[7] = {arr_1d[7]}  # <- modifié aussi car reshape crée une vue")
```

##### 9. Opérations linéaires et matrices

```python
A: NDArray[np.int64] = np.array([[1, 2], [3, 4]])
B: NDArray[np.int64] = np.array([[5, 6], [7, 8]])
```

```python
# Multiplication matricielle avec np.dot() ou @ (pour produit matriciel)
produit_mat: NDArray[np.int64] = np.dot(A, B)
print("Produit matriciel A dot B:\n", produit_mat)

produit_mat = A @ B
print("Produit matriciel A @ B:\n", produit_mat)
```

⚠️ Attention : avec la multiplication element-wise (par exemple A * B) 

Ce n'est pas une multiplication matricielle, mais une multiplication élément par élément.

```python
produit_element_wise: NDArray[np.int64] = A * B
print("Produit élément par élément A * B:\n", produit_element_wise)
```

```python
addition_mat: NDArray[np.int64] = A + B
print("Addition de matrices A + B:\n", addition_mat)
```

L'inverse d'une matrice se calcule avec _np.linalg.inv()_

Attention: l'inverse d'une matrice n'est pas défini si la matrice est singulière (non inversible)

```python
try:
    inverse_A: NDArray[np.float64] = np.linalg.inv(A)
    print("Inverse de la matrice A:\n", inverse_A)
except np.linalg.LinAlgError:
    print("La matrice A n'est pas inversible.")
```

⚠️ Attention avec la division element-wise (par exemple A / B) 

Ce n'est pas la division matricielle, mais une division élément par élément.

```python
division_elementwise: NDArray[np.float64] = A / B
print("Division élément par élément A / B:\n", division_elementwise)
```

```python
transpose_A: NDArray[np.int64] = A.T
print("Transposée A.T:\n", transpose_A)

transpose_A = np.transpose(A)
print("Transposée np.transpose(A):\n", transpose_A)
```

⚠️ Attention : _transpose_ et _.T_ ne retourne pas une copie du tableau, mais une vue (ou perspective) sur les mêmes données.

##### 12. Rééchantillonage

- Changer la fréquence temporelle des données
- Exemple : passer de données horaires à des moyennes journalières

```python
# Création d'une série temporelle quotidienne
dates: pd.DatetimeIndex = pd.date_range('2025-07-27', periods=10, freq='D')
ds_timeseries: pd.Series = pd.Series([10, 12, 11, 15, 14, 13, 16, 18, 17, 20], index=dates)

print("Données quotidiennes originales :")
print(ds_timeseries)

# Rééchantillonnage en mensuel (fin de mois), en prenant la moyenne
# 'ME' : fin de mois ; 'MS' : début de mois
ds_monthly_mean: pd.Series = ds_timeseries.resample('ME').mean()

print("\nDonnées rééchantillonnées en mensuel (moyenne en fin de mois) :")
print(ds_monthly_mean)
```

##### 13. Interpolations

- Compléter les valeurs manquantes
- Exemple : remplir les trous laissés par un rééchantillonnage ou des données absentes



```python
# Rééchantillonnage en intervalles de 12 heures, en prenant le minimum
ds_timeseries_12h: pd.Series = ds_timeseries.resample('12h').min()

print("Données rééchantillonnées toutes les 12 heures (avec valeurs manquantes) :")
print(ds_timeseries_12h[:10])

# Interpolation des valeurs manquantes
ds_interpolees_12h: pd.Series = ds_timeseries_12h.interpolate()

print("\nDonnées interpolées toutes les 12 heures :")
print(ds_interpolees_12h[:10])
```

##### 14. Chaînage (Method Chaining)

Il est possible d’enchaîner plusieurs méthodes pandas sur une même ligne.

Pour améliorer la lisibilité, il est recommandé de répartir le chaînage sur plusieurs lignes
en entourant l'expression avec des parenthèses.

```python
df_chain: pd.DataFrame = (
    df_people.query("Age > 18 and Age < 100") 
    # pd.cut : discrétise la colonne 'Age' en intervalles (bins) avec des étiquettes (labels)
    .assign(age_category=lambda d: pd.cut(
        d["Age"],
        bins=[18, 25, 50, 100],
        labels=["Jeune", "Adulte", "Senior"]
    ))
    .reset_index(drop=True)
    .set_index('Nom')
)

print(df_chain)
```
