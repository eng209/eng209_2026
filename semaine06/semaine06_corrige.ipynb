## Visualisation avec Matplotlib

Les fonctions de pandas sont parfaites pour débuter, mais elles s’appuient sur Matplotlib en arrière-plan. Dans la suite de cet exercice, nous apprendrons à personnaliser davantage nos graphiques avec Matplotlib.

Pour cela, nous allons simuler des valeurs aléatoires tirées d’une loi normale, afficher leur distribution empirique, et tracer la distribution théorique de cette loi pour comparaison.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from numpy.typing import NDArray
```

Definition de la loi de probabilité normale $y={1 \over \sigma\sqrt{2\pi}}e^{{(x-\mu)^2\over 2\sigma^2}} $

```python
def loi_normale(x: NDArray[np.float64], mu: float, sigma: float) -> NDArray[np.float64]:
    return ((1 / (np.sqrt(2 * np.pi) * sigma)) * np.exp(-0.5 * (1 / sigma * (x - mu))**2))
```

Générez 5000 valeurs aléatoires issues d’une distribution normale de moyenne $\mu = 200$ et d’écart-type $\sigma = 25$.

Consultez le notebook [semaine06_antiseche.ipynb](semaine06_antiseche.ipynb) pour voir comment générer ces valeurs aléatoires.

```python
np.random.seed(20221031)

mu = 200
sigma = 25

# Générez 5000 valeurs aléatoires
x: NDArray[np.float64] = np.random.normal(mu, sigma, size=5000)
```

Visualisation l'histogramme de densité des valeurs aléatoires générées à l'aide de la fonction [`matplotlib.pyplot.hist`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html?highlight=hist#matplotlib.pyplot.hist)

Explications:
- figsize=(8,4): définit la taille de la figure (largeur × hauteur en pouces).

- ax.grid(True): ajoute une grille pour faciliter la lecture.

- ax.legend(loc='right'): affiche la légende à droite des axes.

- ax.set_title('Histogramme'): définit le titre du graphique.

- ax.set_xlabel('Valeur') / ax.set_ylabel('Probabilité'): étiquette les axes x et y.

- ax.text(125, .015, r'$\mu=200, \sigma=25$'): ajoute un texte annoté aux coordonnées (125, 0.015) avec mise en forme LaTeX.

- ax.set_xlim(100, 300): fixe les limites de l’axe des x.

- ax.set_ylim(0, 0.02): fixe les limites de l’axe des y.

```python
num_bins=50
fig, ax = plt.subplots(figsize=(8, 4))

bins_y, bins_x, patches = ax.hist(x, num_bins,
                            density=True,
                            cumulative=False, 
                            histtype='bar',
                            label='Empirique')

ax.grid(True)
ax.legend(loc='right')
ax.set_title('Histogramme')
ax.set_xlabel('Valeur')
ax.set_ylabel('Probabilité')

ax.text(125, .015, r'$\mu=200,\ \sigma=25$')
ax.set_xlim(100, 300)
ax.set_ylim(0, 0.02)

plt.show()
```

Modifier l’histogramme ci-dessus pour y superposer la courbe théorique de la loi normale.

💡 Astuces :

- _bin_x_ retourné par la fonction _hist_ correspond aux bords des bins. Sa longueur est _nbins + 1_ (les nbins bords gauche et le bord droit du dernier bin). Ces valeurs peuvent être utilisées pour calculer la densité de la loi normale et la superposer à l’histogramme.

- L’objet _ax_ retourné par la fonction _subplots_ correspond aux axes de la figure, sur lesquels il est possible d’utiliser la méthode plot de Matplotlib pour tracer des courbes supplémentaires, e.g. _ax.plot(x,y, ...)_

- Embellissez votre courbe théorique de la loi normale, en jouant avec les paramètres marker ('+', '.', '^', ...), linestyle ('-', ':', '--', ...), linewidth, etc. Donnez lui un nom.

```python
y: NDArray[np.float64] = loi_normale(bins_x, mu, sigma)
```

```python
# Créer l'espace pour la figure
fig, ax = plt.subplots(figsize=(8, 4))

# Diagramme de l'histogramme des valeurs aléatoires
bins_y, bins_x, patches = ax.hist(x, num_bins,
                           density=True,
                           cumulative=False,
                           histtype='bar',
                           label='Empirique')

# Diagramme théorique de la loi de distribution normale pour mu=200 et sigma=25
ax.plot(bins_x, y, marker='+', linestyle='--', linewidth=1.75, label='Théorique (loi normale)')

# tidy up the figure
ax.grid(True)
ax.legend(loc='right')
ax.set_title('Histogramme')
ax.set_xlabel('Valeur')
ax.set_ylabel('Probabilité')

ax.text(125, .015, r'$\mu=200,\ \sigma=25$')
ax.set_xlim(100, 300)
ax.set_ylim(0, 0.02)

plt.show()
```

## Visualisation 3D avec Matplotlib

Aucun code n’est à écrire pour cet exercice.

Prenez le temps de **comprendre ce que fait le code ci-dessous** et comment chaque étape contribue à la visualisation 3D des données simulées.


```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import colormaps as cm

np.random.seed(20251027)

n = 20000
x = np.random.randn(n)
y = np.random.randn(n)

top, xx, yy = np.histogram2d(x , y, density=True, bins=20, range=[[-4,4],[-4,4]])

xm,ym=np.meshgrid(xx[:-1]+0.25,yy[:-1]+0.25)


fig = plt.figure(figsize=(8,8))
ax  = fig.add_subplot(projection="3d")
ax.grid(True)

ax.plot_surface(xm, ym, top, shade=True, cmap=cm.get_cmap('plasma'))

plt.show()
```

**Explication du code**:

- _np.random.randn(n)_: génère **n valeurs aléatoires** suivant une loi normale centrée réduite pour x et y.  
- _np.histogram2d(x, y, bins=20, range=[[-4,4],[-4,4]], density=True)_: calcule un **histogramme 2D** normalisé, renvoyant :  
  - _top_ : les **valeurs de densité** pour chaque bin,  
  - _xx_ et _yy_ : les **bords des bins** sur les axes x et y.  
- _np.meshgrid(xx[:-1]+0.25, yy[:-1]+0.25)_: construit une **grille 2D** (voir antisèches) des centres des bins pour le tracé de surface.  
- _fig.add_subplot(projection="3d")_: crée des **axes 3D** pour la figure.  
- _ax.plot_surface(xm, ym, top, shade=True, cmap=cm.get_cmap('plasma'))_: trace la **surface 3D**, colorée selon la densité, avec le style _plasma_.  
- _ax.grid(True)` et `figsize=(8,8)_: ajoutent **grille et taille de figure** pour améliorer la lisibilité.  
- _%matplotlib widget_: active l’**interactivité** dans le notebook (rotation, zoom).
