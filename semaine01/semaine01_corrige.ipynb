## Exercice 1: Arithmétique des nombres entiers & interpolation dans les strings

 1. Enlever le `f` fait que les expressions entre accolades ne sont plus interprétées.
 2. Il faut utiliser une concaténation de strings avec l'opérateur `+` et convertir les opérandes qui ne sont pas de type `str` avec la fonction `str(...)`.
 3. En C++, il n'y a qu'un seul opérateur de division (`/`). Le résultat est un nombre entier seulement si ses deux opérandes sont des nombres entiers aussi. Le type de la valeur évaluée dépend donc du type des entrées. En Python, dès la version 3 du langage, `/` calcule toujours un `float` (même si la partie décimale du résultat est zéro) et `//` effectue toujours une division euclidienne. C'est donc l'opérateur qui détermine de manière non ambiguë le type de retour.
 4. Exemple de réponse:

```python
distance = 10
speed = 3
duration = distance // speed

print(f"Distance à parcourir: {distance} km")
print(f"Vitesse: {speed} km/h")

duration_minutes = int(((distance / speed) - duration) * 60)
print(f"Durée du trajet: {duration} h {duration_minutes} min")

# or:
duration_hours, rest = divmod(distance, speed)
duration_minutes = int(rest / speed * 60)
print(f"Durée du trajet: {duration_hours} h {duration_minutes} min")
```

## Exercice 2: Manipulation de chaînes de caractères

```python
name = input("Quel est votre nom? ")
print(name.upper())
print(name.capitalize()) # oui... c'est un peu facile comme ça
# or:
print(name[0].upper() + name[1:].lower())
# lorsqu'on omet le second index, cela veut dire "va jusqu'à la fin"

mid = len(name) // 2  # il faut le double slash pour avoir un int comme résultat, sinon on ne peut pas l'utiliser comme index. Ceci est donc la division euclidienne (donc par exemple 7 / 2 vaut 3)
print(name[0:mid] + '-' + name[mid:])
```

## Exercice 3: Boucles et fonctions

```python
import math

# version OK mais un peu naïve
def is_prime(n: int) -> bool:
    if n == 2:
        return True

    for m in range(2, n):
        if n % m == 0:
            # m est un diviseur de n
            return False
    return True

# version plus optimisée
def is_prime_optimized(n: int) -> bool:
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    square_root = int(math.sqrt(n) + 1)
    for m in range(3, square_root + 1, 2):
        if n % m == 0:
            # m est un diviseur de n
            return False
    return True

print("2 est premier? " + str(is_prime(2)))
print("3 est premier? " + str(is_prime(3)))
print("29 est premier? " + str(is_prime(29)))
print("4 est premier? " + str(is_prime(4)))
print("9 est premier? " + str(is_prime(9)))
print("27 est premier? " + str(is_prime(27)))

# plus concis, on verra cette notation plus tard:
for i in 2, 3, 29, 4, 9, 27:
    print(f"{i} est premier? {is_prime(i)}")

def show_first_primes(n: int) -> None:
    num_found = 0
    i = 2
    while num_found < n:
        if is_prime(i):
            print(i)
            num_found += 1
        i += 1

show_first_primes(50)
```

## Exercice 4: Compréhension de code

Exécutez les deux cellules suivantes:

```python
def show_word(word: str) -> None:
    print(f"Mot: '{word}'")
```

```python
my_string = "This is  fun!"

last_space_position = -1
while (next_space := my_string.find(" ", last_space_position + 1)) != -1:
    word = my_string[last_space_position + 1:next_space]
    show_word(word)
    last_space_position = next_space

last_word = my_string[last_space_position + 1:]
show_word(last_word)
```

 1. Ce programme intère à travers tous les mots de `my_string`, «mot» étant les caractères compris entre un espace et l'espace suivant.
 2. Pour ne pas imprimer les mots vides, on pourrait simplement modifier `show_word` ainsi (pour le tester, exécutez la cellule ci-dessous, puis réexécutez la cellule de code ci-dessus, qui utilisera la nouvelle définition):

```python
def show_word(word: str) -> None:
    if len(word) != 0:
        print(f"Mot: '{word}'")
```
