#### Apprentissage du modèle

Rappel : Il faut toujours valider le modèle sur des données de test distinctes de celles utilisées lors de l’entraînement.

Cette étape est essentielle pour vérifier que le modèle ne s’est pas contenté de mémoriser les données d’entraînement (suroptimisation) et qu’il sait bien généraliser à de nouvelles données.

````python
from numpy.typing import NDArray
from sklearn.model_selection import train_test_split

X: NDArray[np.float64] = seascout_tx_df[["salinity","temperature","depth"]].to_numpy()
y: NDArray[np.float64] = seascout_tx_df.velocity.to_numpy()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=42)
````

Notez le paramètre _random_state_. Avec _train_test_split_, la division entre l’ensemble d’entraînement et l’ensemble de test est faite de façon aléatoire. En fixant _random_state_ à un entier, vous assurez que cette division reste identique en cas d' exécutions successives du split.

Il est en effet essentiel de garder cette division inchangée lors de l’optimisation (hyperparamétrisation) du degré de votre modèle, afin d’éviter toute _fuite des données de test_. C’est-à-dire le fait d’utiliser, même partiellement, les données de test pendant la phase d’entraînement.

````python
print(f"Nombre de données d'entrainement: {y_train.shape}")
print(f"Nombre de données de test: {y_test.shape}")
````

L’équation empirique de Kenneth V. Mackenzie indique un polynôme maximum de degré 4 ($T \cdot D^3$). Nous testerons donc des degrés autour de cette valeur.

````python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn import metrics
from sklearn.pipeline import Pipeline

# Pipeline
pipeline: Pipeline = Pipeline(steps=[
    ('polynomial', PolynomialFeatures(degree=3, include_bias=False)),
    ('standardisation', StandardScaler()),
    ('regression', LinearRegression()), # Régression linéaire classique (pas de régularisation)
    #('regression', Lasso(alpha=0.01, max_iter=100000)),   # Régression Lasso
    ]
)


model=pipeline.fit(X_train, y_train)
````

Intercept et coefficients.

Note: [PolynomialFeatures](https://scikit-learn.org/stable/modules/preprocessing.html#polynomial-features) génère toutes les combinaisons polynomiales des variables d’entrée dont le degré est inférieur ou égal au degré maximal spécifié.

Cela comprend toutes les variables $x_i$, puis les produits $x_i x_j$ avec $j \ge i$, $x_i x_j x_k$ avec $k \ge j \ge i$, et ainsi de suite jusqu'à $x_i^d$, pour un total de $\binom{n+d}{d}$ combinations.

Le terme _include_bias=False_ permet de ne pas inclure le terme de degré 0 correspondant à l'intercept, qui, lui, doit être calculé par le modèle de régression.

Donc, pour degré $d=3$ et  $n=3$ variables indépendantes, le modèle comporte 20 paramètres : 19 coefficients pour les variables polynomiales et 1 intercept calculé par le modèle de régression.

````python
print(f"intercept: {model.named_steps["regression"].intercept_}")
print(f"coefficients:\n{model.named_steps["regression"].coef_}")
````

#### Validation du modèle

##### Validation sur données de test

````python
y_test_pred: NDArray[np.float64] = np.asarray(model.predict(X_test))
````

**Calcule du MAE et RMSE**

````python
print(f"MAE: {metrics.mean_absolute_error(y_test_pred, y_test)}")
print(f"RMSE: {np.sqrt(metrics.mean_squared_error(y_test_pred, y_test))}")
````

**Étude des résiduels**

````python
from scipy.stats import norm
residuals = y_test_pred - y_test
mean_pdf, std_pdf = norm.fit(residuals)
x_pdf = np.arange(-0.1, 0.1, 0.001)

fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(x=residuals, density=True, alpha=0.7, edgecolor='black', color='skyblue', label="Histogramme de densité")
ax.plot(x_pdf, norm.pdf(x_pdf, mean_pdf, std_pdf), color='red', label="Loi normale")
ax.legend(loc='best', fontsize=10)
ax.set_title(r"Distribution de l'erreur $\|\hat{y}-y\|$", fontsize=12)
ax.set_xlabel(r"Erreur $\hat{y}=f(X)$", fontsize=10)
ax.set_yscale('log')
ax.set_xlim(-0.1, 0.1)
ax.set_ylabel("Densité", fontsize=10)
ax.tick_params(axis='both', labelsize=10) 
ax.text(0.01, 40, f"(mean={mean_pdf:2.2f}, std={std_pdf:2.4f})", color='red', fontsize=10)

plt.show()
````

Réponse:

- Le modèle entraîné uniquement sur les données du Texas (TX) ne généralise pas bien aux nouvelles données de Floride (FL) : ses performances sont nettement inférieures à celles obtenues sur TX, car il est trop spécifique aux premières données.

````python
from scipy.stats import norm
residuals_fl = y_pred_test_fl - seascout_fl_df.velocity
mean_pdf, std_pdf = norm.fit(residuals_fl)
x_pdf = np.arange(-0.4, 0.4, 0.001)

fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(x=residuals_fl, density=True, alpha=0.7, edgecolor='black', color='skyblue', label="Histogramme de densité")
ax.plot(x_pdf, norm.pdf(x_pdf, mean_pdf, std_pdf), color='red', label="Loi normale")
ax.legend(loc='best', fontsize=10)
ax.set_title(r"Distribution de l'erreur $\|\hat{y}-y\|$", fontsize=12)
ax.set_xlabel(r"Erreur $\hat{y}=f(X)$", fontsize=10)
ax.set_yscale('log')
ax.set_xlim(-0.4, 0.4)
ax.set_ylabel("Densité", fontsize=10)
ax.tick_params(axis='both', labelsize=10) 
ax.text(0.01, 40, f"(mean={mean_pdf:2.2f}, std={std_pdf:2.4f})", color='red', fontsize=10)

plt.show()
````

La moyenne des résidus n’est pas nulle, ce qui confirme notre remarque précédente : ce modèle doit être réentraîné en tenant compte de ces nouvelles données.
